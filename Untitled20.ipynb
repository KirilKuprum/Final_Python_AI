```python
import re, warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.base import clone
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.dummy import DummyRegressor
from sklearn.impute import SimpleImputer
from sklearn.inspection import permutation_importance
from sklearn.linear_model import Ridge, Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import (GridSearchCV, KFold, RepeatedKFold, cross_val_predict,
                                     cross_validate, learning_curve, train_test_split)
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeRegressor

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
pd.options.display.float_format = "{:,.2f}".format
RANDOM_STATE = 42
```

```python
DATA_PATH = Path("i_kvartal_2026.xlsx")
if not DATA_PATH.exists():
    found = sorted(Path(".").glob("*kvartal*.xlsx"))
    if not found:
        raise FileNotFoundError("Покладіть файл i_kvartal_2026.xlsx поруч із ноутбуком")
    DATA_PATH = found[0]

raw_fines = pd.read_excel(DATA_PATH, sheet_name=0, skiprows=[1])
raw_wim = pd.read_excel(DATA_PATH, sheet_name=1)
print("Лист 1:", raw_fines.shape, "| Лист 2:", raw_wim.shape)
raw_fines.tail(4)
```

Завантаження даних


```python
def region_key(text):
    """Спільний ключ для обох листів: перші 6 літер назви області
    ('...у Вінницькій області' і 'Вінницька' -> 'вінниц')."""
    m = re.search(r"\)\s*у\s+(\S+)", str(text))
    word = m.group(1) if m else str(text)
    return word.lower()[:6]

totals = raw_fines[raw_fines["territorialAuthorities"] == "Усього"].iloc[0]

fines = raw_fines[raw_fines["identifier"].notna()].copy()
fines["region"] = fines["territorialAuthorities"].str.extract(r"\)\s*у\s+(\S+)")[0]
fines["key"] = fines["territorialAuthorities"].map(region_key)

wim = raw_wim[~raw_wim["wimAddressAdminUnitL2"].isin(["Всього", "Сплачено до бюджету"])].copy()
wim = wim.rename(columns={"total number of resolutions": "wim_resolutions", "fineAmount": "wim_fine_amount"})
wim["key"] = wim["wimAddressAdminUnitL2"].map(region_key)

MACRO = {"волинс": "Захід", "закарп": "Захід", "івано-": "Захід", "львівс": "Захід", "рівнен": "Захід",
         "терноп": "Захід", "хмельн": "Захід", "чернів": "Захід",
         "вінниц": "Центр", "житоми": "Центр", "київсь": "Центр", "кірово": "Центр", "полтав": "Центр", "черкас": "Центр",
         "чернігі": "Північ", "сумськ": "Північ",
         "дніпро": "Схід", "харків": "Схід", "запорі": "Схід",
         "микола": "Південь", "одеськ": "Південь", "херсон": "Південь"}
MACRO = {k[:6]: v for k, v in MACRO.items()}
fines["macro_region"] = fines["key"].map(MACRO)

df = fines.merge(wim[["key", "wim_resolutions", "wim_fine_amount"]], on="key", how="left")
df = df[["region", "macro_region", "totalAmountFines", "finesTruckTransport", "finesPassengerTransport",
         "paidFines", "paidFinesTruck", "paidFinesPassenger", "wim_resolutions", "wim_fine_amount"]].reset_index(drop=True)

assert df["macro_region"].notna().all(), "є область без макрорегіону"
assert (df["finesTruckTransport"] + df["finesPassengerTransport"] == df["totalAmountFines"]).all()
print("Рядків (управлінь):", len(df))
print("Сума накладених штрафів по управліннях:", f"{df['totalAmountFines'].sum():,.0f}",
      "| у підсумку файлу (разом із відділом оперативного реагування):", f"{totals['totalAmountFines']:,.0f}")
df
```

Кореляційний аналіз

```python
def make_preprocessor():
    num = Pipeline([("imp", SimpleImputer(strategy="median")),
                    ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one")),
                    ("sc", StandardScaler())])
    cat = Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
                    ("oh", OneHotEncoder(handle_unknown="ignore"))])
    return ColumnTransformer([("num", num, NUM), ("cat", cat, CAT)])

def make_model(estimator):
    pipe = Pipeline([("prep", make_preprocessor()), ("est", estimator)])
    return TransformedTargetRegressor(regressor=pipe, func=np.log1p, inverse_func=np.expm1)

X, y = df[NUM + CAT], df[TARGET]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=RANDOM_STATE)
print(f"train: {len(X_train)} записів, test: {len(X_test)} записів")
```

Налаштування пайплайна

```python
MODELS = {
    "Ridge": (Ridge(), {"regressor__est__alpha": [0.01, 0.1, 1, 10, 100]}),
    "Lasso": (Lasso(max_iter=10000), {"regressor__est__alpha": [0.001, 0.01, 0.05, 0.2]}),
    "Дерево рішень": (DecisionTreeRegressor(random_state=RANDOM_STATE),
                      {"regressor__est__max_depth": [2, 3, 4, None], "regressor__est__min_samples_leaf": [1, 2, 3]}),
    "MLPRegressor": (MLPRegressor(solver="lbfgs", max_iter=5000, random_state=RANDOM_STATE),
                     {"regressor__est__hidden_layer_sizes": [(8,), (16,), (8, 8)],
                      "regressor__est__alpha": [0.01, 1.0]}),
}
inner_cv = RepeatedKFold(n_splits=4, n_repeats=2, random_state=RANDOM_STATE)
outer_cv = RepeatedKFold(n_splits=4, n_repeats=3, random_state=0)
SCORING = {"mae": "neg_mean_absolute_error", "rmse": "neg_root_mean_squared_error", "r2": "r2"}

def test_metrics(model):
    p = model.predict(X_test)
    return (mean_absolute_error(y_test, p), np.sqrt(mean_squared_error(y_test, p)), r2_score(y_test, p))

rows, fitted = [], {}

base = DummyRegressor(strategy="mean").fit(X_train, y_train)
nb = cross_validate(DummyRegressor(strategy="mean"), X, y, cv=outer_cv, scoring=SCORING)
mae_t, rmse_t, r2_t = test_metrics(base)
rows.append({"Модель": "Базовий рівень (середнє)", "Найкращі параметри": "—",
             "CV MAE, грн": -nb["test_mae"].mean(), "± std": nb["test_mae"].std(),
             "CV R²": nb["test_r2"].mean(), "Test MAE": mae_t, "Test RMSE": rmse_t, "Test R²": r2_t})

for name, (est, grid) in MODELS.items():
    gs = GridSearchCV(make_model(est), grid, cv=inner_cv, scoring="neg_mean_absolute_error", n_jobs=-1)
    gs.fit(X_train, y_train)
    fitted[name] = gs
    nested = cross_validate(GridSearchCV(make_model(est), grid, cv=inner_cv, scoring="neg_mean_absolute_error"),
                            X, y, cv=outer_cv, scoring=SCORING, n_jobs=-1)
    mae_t, rmse_t, r2_t = test_metrics(gs)
    rows.append({"Модель": name,
                 "Найкращі параметри": {k.split("__")[-1]: v for k, v in gs.best_params_.items()},
                 "CV MAE, грн": -nested["test_mae"].mean(), "± std": nested["test_mae"].std(),
                 "CV R²": nested["test_r2"].mean(), "Test MAE": mae_t, "Test RMSE": rmse_t, "Test R²": r2_t})

results = pd.DataFrame(rows).set_index("Модель")
pd.options.display.float_format = "{:,.2f}".format
display(results.sort_values("CV MAE, грн"))
```

Навчання моделей та крос-валідація

```python
order = results.sort_values("CV MAE, грн")
fig, ax = plt.subplots(figsize=(9, 3.8))
ax.barh(order.index[::-1], order["CV MAE, грн"][::-1], xerr=order["± std"][::-1], color="#3b7ea1", capsize=3)
ax.set(title="Вкладена крос-валідація: MAE, грн (менше — краще)", xlabel="MAE, грн")
plt.tight_layout(); plt.show()

BASE = "Базовий рівень (середнє)"
COMPLEXITY = ["Ridge", "Lasso", "Дерево рішень", "MLPRegressor"]
cand = results.drop(index=BASE)
top = cand["CV MAE, грн"].idxmin()
threshold = cand.loc[top, "CV MAE, грн"] + cand.loc[top, "± std"]
best_name = next(n for n in COMPLEXITY if cand.loc[n, "CV MAE, грн"] <= threshold)
print(f"Найменший CV MAE: {top} ({cand.loc[top, 'CV MAE, грн']:,.0f} грн), поріг «не гірше за 1 std»: {threshold:,.0f} грн")
print("Обрана модель (найпростіша в межах порогу):", best_name)
```

Порівняння і вибір моделі

```python
est, grid = MODELS[best_name]
final_search = GridSearchCV(make_model(est), grid, cv=inner_cv, scoring="neg_mean_absolute_error", n_jobs=-1).fit(X, y)
final_model = final_search.best_estimator_
print("Фінальна модель:", best_name, "| параметри:", {k.split("__")[-1]: v for k, v in final_search.best_params_.items()})

NESTED_MAE = results.loc[best_name, "CV MAE, грн"]
NESTED_R2 = results.loc[best_name, "CV R²"]
BASE_MAE = results.loc["Базовий рівень (середнє)", "CV MAE, грн"]
print(f"Очікувана похибка (nested CV): MAE ≈ {NESTED_MAE:,.0f} грн; базовий рівень: {BASE_MAE:,.0f} грн; R² ≈ {NESTED_R2:.2f}")
```

Збереження моделі

```python
bundle = joblib.load(MODEL_PATH)
REQUIRED = {"finesTruckTransport": "Штрафи, вантажний транспорт, грн",
            "finesPassengerTransport": "Штрафи, пасажирський транспорт, грн"}
OPTIONAL = {"wim_resolutions": "Постанови WIM, шт", "wim_fine_amount": "Штрафи WIM, грн"}

def _to_number(v):
    if v is None or (isinstance(v, str) and not v.strip()):
        return None
    return float(str(v).replace(" ", "").replace(",", "."))

def validate_input(raw, bundle=bundle):
    """Повертає (чисті значення, список помилок, список попереджень)."""
    errors, warns, clean = [], [], {}
    for col, label in {**REQUIRED, **OPTIONAL}.items():
        try:
            v = _to_number(raw.get(col))
        except ValueError:
            errors.append(f"«{label}»: потрібне число"); continue
        if v is None:
            if col in REQUIRED:
                errors.append(f"«{label}»: поле порожнє")
            else:
                warns.append(f"«{label}» не вказано — використано медіану навчальних даних")
            clean[col] = np.nan
            continue
        if not np.isfinite(v) or v < 0:
            errors.append(f"«{label}»: значення має бути невід'ємним числом"); continue
        lo, hi = bundle["ranges"][col]
        if v < lo or v > hi:
            warns.append(f"«{label}» = {v:,.0f} поза діапазоном навчальних даних [{lo:,.0f}; {hi:,.0f}] — прогноз менш надійний")
        clean[col] = v
    region = raw.get("macro_region")
    if region not in bundle["macro_regions"]:
        errors.append(f"Макрорегіон має бути одним із: {', '.join(bundle['macro_regions'])}")
    clean["macro_region"] = region
    return clean, errors, warns

def predict_paid_fines(raw, bundle=bundle):
    clean, errors, warns = validate_input(raw, bundle)
    if errors:
        return {"ok": False, "errors": errors, "warnings": warns}
    row = pd.DataFrame([clean])[bundle["num_features"] + bundle["cat_features"]]
    model = bundle["model"]
    pred = float(model.predict(row)[0])

    contrib = {}
    for col in bundle["num_features"]:
        alt = row.copy(); alt[col] = bundle["medians"][col]
        contrib[col] = pred - float(model.predict(alt)[0])
    alt = row.copy(); alt["macro_region"] = bundle["default_region"]
    contrib["macro_region"] = pred - float(model.predict(alt)[0])

    mae = bundle["expected_mae"]
    return {"ok": True, "prediction": pred, "low": max(0.0, pred - mae), "high": pred + mae,
            "expected_mae": mae, "baseline_mae": bundle["baseline_mae"],
            "contributions": dict(sorted(contrib.items(), key=lambda kv: -abs(kv[1]))),
            "warnings": warns, "errors": []}

def show_result(res):
    if not res["ok"]:
        print("Помилки вводу:"); [print("  •", e) for e in res["errors"]]; return
    print(f"Прогноз суми, перерахованої до бюджету: {res['prediction']:,.0f} грн")
    print(f"Очікувана похибка (nested CV): ±{res['expected_mae']:,.0f} грн  → орієнтовний діапазон "
          f"{res['low']:,.0f} … {res['high']:,.0f} грн")
    print("Внесок ознак (зміна прогнозу порівняно з типовим значенням ознаки):")
    for k, v in res["contributions"].items():
        print(f"  {k:26s} {v:+15,.0f} грн")
    for w_ in res["warnings"]:
        print(w_)

show_result(predict_paid_fines({"finesTruckTransport": 5_000_000, "finesPassengerTransport": 1_000_000,
                                "wim_resolutions": 40, "wim_fine_amount": 500_000, "macro_region": "Центр"}))
print("\n--- некоректний ввід ---")
show_result(predict_paid_fines({"finesTruckTransport": "", "finesPassengerTransport": "abc", "macro_region": "Марс"}))
print("\n--- значення поза діапазоном ---")
show_result(predict_paid_fines({"finesTruckTransport": 900_000_000, "finesPassengerTransport": 1_000_000, "macro_region": "Схід"}))
```

Інтерфейс: форма вводу

```python
if HAVE_WIDGETS:
    r = bundle["ranges"]
    sl = {c: w.FloatSlider(value=bundle["medians"][c], min=r[c][0], max=r[c][1], step=max(1.0, (r[c][1] - r[c][0]) / 200),
                           description=lbl, readout_format=",.0f", continuous_update=False,
                           style={"description_width": "230px"}, layout=w.Layout(width="650px"))
          for c, lbl in {**REQUIRED, **OPTIONAL}.items()}
    dd = w.Dropdown(options=bundle["macro_regions"], description="Макрорегіон", style={"description_width": "230px"})

    def what_if(**kw):
        show_result(predict_paid_fines({**{k: kw[k] for k in sl}, "macro_region": kw["macro_region"]}))

    display(w.interactive(what_if, macro_region=dd, **sl))
```
